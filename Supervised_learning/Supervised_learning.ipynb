## О проекте

**Описание проекта**

Отчёт продающего товары разных категорий (для детей, для дома, мелкую бытовую технику, косметику, продукты) интернет-магазина «В один клик» за прошлый период показал, что активность покупателей начала снижаться. Привлекать новых клиентов уже не так эффективно: о магазине и так знает большая часть целевой аудитории. Возможный выход — удерживать активность постоянных клиентов. Сделать это можно с помощью персонализированных предложений.

«В один клик» — современная компания, поэтому её руководство не хочет принимать решения просто так — только на основе анализа данных и бизнес-моделирования, и она поручила решение задачи своему небольшому отделу цифровых технологий.

**Цели проекта**

Разработать решение, что позволит персонализировать предложения постоянным клиентам для увеличения их покупательской активность.

**Описание решения**  

Руководитель отдела сформировал описание решения для данной задачи, которое одобрило руководство:

1. Промаркировать уровень финансовой активности постоянных покупателей. В компании принято выделять два уровня активности: «снизилась», если клиент стал покупать меньше товаров, и «прежний уровень».

2. Собрать данные по клиентам по следующим группам:
     
       - Признаки, которые описывают коммуникацию сотрудников компании с клиентом.
       - Признаки, которые описывают продуктовое поведение покупателя. Например, какие товары покупает и как часто.
       - Признаки, которые описывают покупательское поведение клиента. Например, сколько тратил в магазине.
       - Признаки, которые описывают поведение покупателя на сайте. Например, как много страниц просматривает и сколько времени проводит на сайте.

3. Построить модель, которая предскажет вероятность снижения покупательской активности клиента в следующие три месяца.
4. Включить в исследование дополнительные данные финансового департамента о прибыльности клиента: какой доход каждый покупатель приносил компании за последние три месяца.
5. Используя данные модели и данные о прибыльности клиентов, выделить сегменты покупателей и разработать для них персонализированные предложения.

**Описание данных**

Мы будем работать с четырьмя датасетами.

1. `'market_file.csv'`
Таблица, которая содержит данные о поведении покупателя на сайте, о коммуникациях с покупателем и его продуктовом поведении. Включает в себя следующие столбцы:

- `'id'` — номер покупателя в корпоративной базе данных.
- `'Покупательская активность'` — рассчитанный класс покупательской активности (целевой признак): «снизилась» или «прежний уровень».
- `'Тип сервиса'` — уровень сервиса, например «премиум» и «стандарт».
- `'Разрешить сообщать'` — информация о том, можно ли присылать покупателю дополнительные предложения о товаре. Согласие на это даёт покупатель.
- `'Маркет_актив_6_мес'` — среднемесячное значение маркетинговых коммуникаций компании, которое приходилось на покупателя за последние 6 месяцев. Это значение показывает, какое число рассылок, звонков, показов рекламы и прочего приходилось на клиента.
- `'Маркет_актив_тек_мес'` — количество маркетинговых коммуникаций в текущем месяце.
- `'Длительность'` — значение, которое показывает, сколько дней прошло с момента регистрации покупателя на сайте.
- `'Популярная_категория'` — самая популярная категория товаров у покупателя за последние 6 месяцев.
- `'Средний_просмотр_категорий_за_визит'` — показывает, сколько в среднем категорий покупатель просмотрел за визит в течение последнего месяца.
-`'Неоплаченные_продукты_штук_квартал'` — общее число неоплаченных товаров в корзине за последние 3 месяца.
-`'Ошибка_сервиса'` — число сбоев, которые коснулись покупателя во время посещения сайта.
-`'Страниц_за_визит'` — среднее количество страниц, которые просмотрел покупатель за один визит на сайт за последние 3 месяца.

2. `'market_money.csv'`
Таблица с данными о выручке, которую получает магазин с покупателя, то есть сколько покупатель всего потратил за период взаимодействия с сайтом.

- `'id'` — номер покупателя в корпоративной базе данных.
- `'Период'` — название периода, во время которого зафиксирована выручка. Например, 'текущий_месяц' или 'предыдущий_месяц'.
- `'Выручка'` — сумма выручки за период.

3. `'market_time.csv'`
Таблица с данными о времени (в минутах), которое покупатель провёл на сайте в течение периода.

- `'id'` — номер покупателя в корпоративной базе данных.
- `'Период'` — название периода, во время которого зафиксировано общее время.
- `'минут'` — значение времени, проведённого на сайте, в минутах.

4. `'money.csv'`
Таблица с данными о среднемесячной прибыли покупателя за последние 3 месяца: какую прибыль получает магазин от продаж каждому покупателю.

- `'id'` — номер покупателя в корпоративной базе данных.
- `'Прибыль'` — значение прибыли.

## Пайплайны для работы с моделями

Чтобы не повторять раз за разом один и тот же код, а также не обучать и не выбирать модель вручную, удобнее всего использовать пайплайны.

Это позволит нам применить сразу четыре модели: k-ближайших соседей (KNeighborsClassifier()), дерево решений (DecisionTreeClassifier()), логистическую регрессию (LogisticRegression()) и метод опорных векторов (SVC()); и произвести разные варианты их настроек.

Метрикой, на которую мы будем ориентироваться, будет `'ROC-AUC'`: площадь под ROC-кривой, которая показывает зависимость TPR(доля верно классифицированных объектов класса 1) и FPR (доля неверно классифицированных объектов класса 0). У целевого признака (`'покупательская активность'`) всего два класса (не мультиклассовая классификация): `'снизилась'` (класс 1) и `'прежний уровень'` (класс 0), и оба важно классифицировать правильно (так как ошибочно предсказанное снижение - потери на бонусных предложениях, ошибочно предсказанное сохранение уровня - убытки от снижения активности непростимулированного покупателя). Она не зависит от дисбаланса классов и выбранного порога вероятности (оценивает качество решений модели в наиболее общем виде, учитывая все возможные пороговые значения) и применима к разным моделям. Поэтому именно данная метрика позволит правильно оценить качество работы разных моделей как на имеющихся данных, так и на будущих.

Теперь можно приступать к сбору пайплайна.

Сначала добавляем  в пайплайн OHE-кодирование. С одной стороны, так как пропусков и неизвестных значений у нас нет, то SimpleImputer нам не требуется и его можно закомментировать. С другой - качество тестовых данных может отличаться от тренировочных и валидационных не в лучшую сторону, поэтому работу с пропусками лучше сохранить.

```python
# SimpleImputer + OHE
ohe_pipe = Pipeline(
    [
        # Если в данных есть пропуски, заполним их самым часто встречающимся значением с помощью SimpleImputer
        (
            'simpleImputer_ohe',
            SimpleImputer(missing_values=np.nan, strategy='most_frequent')
        ),
        # Кодируем данные с помощью OneHotEncoder, избегая дамми-ловушки, игнорируя неизвестные значения,
        # и чтобы результаты кодирования можно было визуально увидеть - не сохраняем его в sparce-формате
        (
            'ohe',
            OneHotEncoder(drop='first', handle_unknown='ignore', sparse=False)
        )
    ]
)
```

Добавляем Ordinal-кодирование в пайплайн.

```python
# SimpleImputer + ORD + SimpleImputer

'''
При порядковом кодировании необходимо задать порядок категоризации.
Поэтому в параметре categories последовательно перечислим
уровни категорий для каждого признака из списка ord_columns
'''

ord_pipe = Pipeline(
    [
        # Заполним пропуски самым часто встречающимся значением с помощью SimpleImputer
        (
            'simpleImputer_before_ord',
            SimpleImputer(missing_values=np.nan, strategy='most_frequent')
        ),
        # Проводим Ordinal-кодирование с помощью OrdinalEncoder
        (
            'ord',
            OrdinalEncoder(
                categories=[
                    # перечисляем от меньшего к большему
                    ['стандарт', 'премиум'],
                    ['менее 50%', '50% и более']
                ],
                # неизвестные значения превращаем в пропуски
                handle_unknown='use_encoded_value', unknown_value=np.nan
            )
        ),
        # Заполним пропуски, которые появляются после Ordinal-кодирования
        (
            'simpleImputer_after_ord',
            SimpleImputer(missing_values=np.nan, strategy='most_frequent')
        )
    ]
)
```

Собираем общий пайплайн для подготовки данных.

```python
# объединяем шаги подготовки
data_preprocessor = ColumnTransformer(
    [
        ('ohe', ohe_pipe, ohe_columns),
        ('ord', ord_pipe, ord_columns),
        ('num', MinMaxScaler(), num_columns) # указываем один скалер из планируемых, далее будем их уточнять
    ],
    remainder='passthrough' # позволяет не трогать данные, не вошедшие в списки выше
)
```

Заготовки для преобразования данных готовы  преобразование целевого признака выполнено. Можно переходить к следующему шагу.

### Итоговый пайплайн: подготовка данных и модель

Теперь мы можем собрать итоговый пайплайн, включив туда подготовку данных и использование моделей, а затем (уже отдельно) дополнительно уточнить варианты применяемого к данным масштабирования и варианты моделей и их параметров.

```python
# создаём итоговый пайплайн: подготовка данных и модель
pipe_final = Pipeline(
    [
        ('preprocessor', data_preprocessor),
        # указываем любую модель из планируемых, так как далее будем их уточнять
        ('models', DecisionTreeClassifier(random_state=RANDOM_STATE))
    ]
)
```

```python
# создаем список скелеров
skel = [StandardScaler(), MinMaxScaler(), RobustScaler(), 'passthrough']
```

```python
# описание моделей и их гиперпараметров
param_grid = [
    # словарь для модели DecisionTreeClassifier()
    {
        'models': [DecisionTreeClassifier(random_state=RANDOM_STATE)],
        'models__max_depth': range(2, 12),
        'models__max_features': range(2, 12),
        'preprocessor__num': skel
    },

    # словарь для модели KNeighborsClassifier()
    {
        'models': [KNeighborsClassifier()],
        'models__n_neighbors': range(2, 10),
        'preprocessor__num': skel
    },

    # словарь для модели SVC(), probability=True нужно для возможности получать вероятность (напр. для pedict_proba и SHAP)
    {
        'models': [SVC(random_state=RANDOM_STATE, probability=True)],
        'models__kernel': ['linear', 'rbf', 'sigmoid'], # варианты ядер
        'models__C': range(1, 10), # сложность модели
        'preprocessor__num': skel
    },

    # словарь для модели LogisticRegression()
    {
        'models': [LogisticRegression(
            random_state=RANDOM_STATE,
            penalty='elasticnet', # регуляризация, смешанный вариант штрафов
            solver='saga' # saga - для l1 и elasticnet
        )],
        'models__C': range(1, 10), # сложность модели
        'preprocessor__num': skel
    }
]
```

### Результат: лучшая модель и ее параметры

Так как данных у нас не слишком много, а также мы ограничили "области поиска" моделям с помощью гиперпараметров, мы можем попробовать использовать GridSearchCV чтобы найти наилучшие из всех возможных вариантов.

```python
# перебор параметров по сетке
grid = GridSearchCV(
    pipe_final,
    param_grid=param_grid,
    cv=5,
    # задаем метрику
    scoring='roc_auc',
    n_jobs=-1
)
```

Наконец мы можем приступить к графику важности признаков.

Больше всего информации может дать метод SHAP. При попытках применения иные методы не сработали. Перейдем к SHAP.

Так как у нас очень много наблюдений, то метод будет оценивать признаки очень долго. Поэтому сократим методу объем работы (а себе - время ожидания сделав выборку из выборки). Пусть мы и потеряем нексолько в точности оценки, но сможем проследить общие тенденции.

```python
background_samples_test_rs = shap.sample(X_test_rs, 30) # выборка из выборки для тестовых данных
```

```python
# объясняем предсказания на тестовой выборке (в нашем случае не все, а на выборке из выборки для экономии времени):
# 1. создаем объект, который содержит информацию о модели машинного обучения и используется для объяснения её предсказаний
explainer_test_rs = shap.KernelExplainer(svc_model_rs.predict_proba, background_samples_test_rs)
```

```python
# 2. вычисляем значения SHAP (SHapley Additive exPlanations) для каждого объекта в тестовом наборе данных.
# Значения SHAP показывают, насколько каждый признак влияет на предсказание модели.
shap_values_test_rs = explainer_test_rs.shap_values(background_samples_test_rs)
```

```python
# 7. Visualize SHAP summary plot for global feature importance
# перевод: визуализация итогогового графика SHAP для глобальной важности признаков
# смотрим по классу 0
shap.summary_plot(shap_values_test_rs[:, :, 0], background_samples_test_rs)
```

Теперь попробуем доработать наш график для удобства восприятия информации и ее интерпретации.

А также интерпритируем исходя из того, что:

•	Негативные значения Шепли слева от центральной вертикальной линии означают, что признак склоняет модель отнести объекты к классу 0, а положительные справа — к классу 1.

•	Чем толще линия по вертикали, тем больше наблюдений сгруппированы вместе: у них близкие значения Шепли. Это значит, что модель уверенно присваивает SHAP-значения, с учётом закономерностей в данных.

•	Цвет точки отражает значения каждого признака объекта: чем краснее объект, тем больше признак, указанный на оси Y.


```python
shap.summary_plot(shap_values_test_rs[:, :, 1], background_samples_test_rs,\
                  feature_names=features_out_rs, plot_type="violin", show=False)

plt.ylabel('важность признаков <---', fontsize=13)
plt.xlabel('вклад в прогноз')
plt.title('Важность признаков и их вклад в прогноз для класса 1 (активность снизилась)\n \
по мнению лучшей из моделей, полученных случайным перебором', fontsize=20)
plt.show()
```

На первом месте по важности оказалось проведенное на сайте время в текущем месяце: чем оно меньше, чем больше шансов на то, что клиент снизит свою активность (это может значить, что наши товары его больше не привлекают по каким-то причинам, ведь если смотрел, но не купил - это одно, может быть ждет акции или зарплату, а если даже не смотрел - дело уже плохо), уже в меньшей степени, но тоже оказывает влияние - снижение проведенного на сайте времени в предыдущем месяце и высокое количество  просмотренных страниц за квартал (что логично, ведь если раньше было много, а сейчас - мало, то произошло снижение).

На то, что клиент снижает свою активность больше всего указывает уменьшение выручки в предыдущем и предпредыдущем месяцах (в меньшей - в текущем, что логично, ведь один месяц это может быть разовой ситуацией, а не тенценцией).

Что тоже логично: раз покупатель активно "сорит деньгами", вряд ли он резко передумает (непредвиденные обстоятельства вроде ограбления, и полупредвидмимые вроде банкротства (мы не всегда можем знать, чьи деньги - свои или кредит от банка - тратит клиент) не в счет, на то они и непредвидимые).

Можно ввести систему поощрения:
- накопление баллов (вероятно, лучше сгорающих, если их не потратить за определенное время, чем "постоянных"; причем которые нельзя потратить сразу, а например, только в следующем месяце, но для этого лучше провести дополнительное исследование.
- скидочную систему для постоянных покупателей с ранжированием в зависимости от потраченной за определенный период суммы (пример - АлиЭкспресс в одной из своих вариций), партнерские программы с иными фирмами (например, купив товара на определенную сумму, получи купон на скидку на посещение аквапарка или на бесплантую пиццу в пиццерии-партнере, а они взамен, при покупке чего-то у них давали бы купоны на скидки у нас).
- скидки на следующую покупку при покупке на определенную сумму сейчас, причем можно также сделать, что скидка сгорает, если ее не использовать за определенный период.
- условный кэшбэк. Как пример: ГлорияДжинс, которая начисляет на картулояльности до 50% баллами от стоимости товаров по полной цене для оплаты до 50% следующие покупки, который можно потратить в течении 90 дней и сгорает, если не использоуется. При этом не факт, что человеку в этот понадобиться еще одна вещь, но ему будет приятно, и может простимулировать его купить еще, даже если и не очень было надо, либо привести друга/родственника, который купит что-то уже себе. Так как наши товары представлены в более широком ассортименте категорий (не только одежда, обувь, аксессуары, как в примере, то есть то, что не слишком часто покупается), то шанс возвращения клиента за чем-нибудь еще выше, поэтому скидка все-таки должна быть меньше чем 50% (а какая оптимальная - подскажет дополнительный анализ), либо разные начесления и скидки по разным категориям (например больше начисления - по дорогим и более редко покупаемым товарам, на акционные или стоимочтью до определенной суммы начисление не производится),
- промокоды на скидки (их, в отличие от персональной скидки, можно передать кому-то другому).

Но так как мы магазин для широкого круга покупателей, модель с эксклюзивными товарами для особых категорий (категорий - полученных по сумме трат за определенный период, условных "бриллиантовых" и "золотых") на не подходит. Обычного покупателя (которых тут большинство) такое скорее оттолкнет ("жмоты, для буржуев держите"), чем включит азарт и желание получить такие вещи (такое подошло бы для магазина с элитными товарами или клубов коллекционеров, например, у того же Mattel, где лишь члены их клуба могут принимать участия в некоторых голосованиях на выбор ново модели куклы, а также первоочереное (а иногда и эксклюзивное) право на покупку некоторых из моделей).

Но в любом случае,чем удобнеее и стабильнее сайт, тем приятнее им пользоваться. Ведь какими бы ни были приятными цены и ассортимент, если он будет неудобным, непонятным и глючащим - покупатель постарается уйти туда, где ему будет удобнее.

Популярность за полгода товаров из "Мелкой бытовая техника и электроника" ("целиком" популярные категории за полгода показали умеренную корреляцию) модель от случайного пребора поставила на пятое место, а вот модель от перебора по сетке поставила лишь на восьмое. Возможно, вторая "сообразила", свой вклад в активность клиента данная категория может вносить своей часто высокой стоимостью, а потому - скурпулезности и при выборе (то есть необходимость изучить много страниц, для чего провести на сайте большое количество времени, а потом потратить крупную сумму). Такие покупки редки, поэтому возможно модель от перебора по сетке не сочла их настолько важными, как модель от случайного перебора.

Остальные категории популярных товаров практически не влияют на картину (похоже, не смотря на умеренную связь по мнению матрицы корреляции они не влияют в приниципе, просто техника связано с высокой ценой и необходимостью более ответственного подхода к выбору и покупке).

Но если оно уже стало тенденцией (та же выручка предпредыдущего и предыдущего месяца, которые снижаются обе) - это плохой знак. Рекомендованные меры - аналогичные "родственным" им признакам.

#### SHAP от RandomizedSearchCV()

Применим SHAP к нашей выборке.

```python
# трансформируем данные
target_segm_f_rs = pd.DataFrame(best_model_rs.named_steps['preprocessor']\
                                  .transform(target_segm_f.drop(['прибыль'],axis=1)),
                                 columns=features_out_rs
                                 )
```

```python
# объясняем предсказания на сегменте. не очень много объектов, то можно обойтись без выбоки из выборки:
# 1. создаем объект, который содержит информацию о модели машинного обучения и используется для объяснения её предсказаний
explainer_segm_f_rs = shap.KernelExplainer(svc_model_rs.predict_proba, target_segm_f_rs)
```

```python
# 2. вычисляем значения SHAP (SHapley Additive exPlanations) для каждого объекта в тестовом наборе данных.
# Значения SHAP показывают, насколько каждый признак влияет на предсказание модели.
shap_values_segm_f_rs = explainer_segm_f_rs.shap_values(target_segm_f_rs)
```

```python
shap.summary_plot(shap_values_segm_f_rs[:, :, 1], target_segm_f_rs,\
                  feature_names=features_out_rs, plot_type="violin", show=False)

plt.ylabel('важность признаков <---', fontsize=13)
plt.xlabel('вклад в прогноз')
plt.title('Важность признаков и их вклад в прогноз для класса 1 (активность снизилась)\n \
по мнению лучшей из моделей, полученных случайным перебором', fontsize=20)
plt.show()
```

Общая картина похожа: три лидера прежние (проведенное на сайте время в текущем месяце и выручки за предыпредыдущий и предыдущие месяцы). И аутсайдеры те же: категории (кроме бытовой техники) и маркетинговая активность по-прежнему оказывают слабое влияние на активность.
