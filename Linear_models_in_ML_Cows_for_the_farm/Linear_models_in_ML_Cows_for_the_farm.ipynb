Файл `'ferma_main.csv'` содержит данные о стаде фермера на текущий момент.

Описание данных:

- `'id'` — уникальный идентификатор коровы.
- `'Удой, кг'` — масса молока, которую корова даёт в год (в килограммах).
- `'ЭКЕ (Энергетическая кормовая единица)'` — показатель питательности корма коровы.
- `'Сырой протеин, г'` — содержание сырого протеина в корме (в граммах).
- `'СПО'` (Сахаро-протеиновое соотношение) — отношение сахара к протеину в корме коровы.
- `'Порода'` — порода коровы.
- `'Тип пастбища'` — ландшафт лугов, на которых паслась корова.
- `'порода папы_быка'` — порода папы коровы.
- `'Жирность,%'` — содержание жиров в молоке (в процентах).
- `'Белок,%'` — содержание белков в молоке (в процентах).
- `'Вкус молока'` — оценка вкуса по личным критериям фермера, бинарный признак (вкусно, не вкусно).
- `'Возраст'` — возраст коровы, бинарный признак (менее_2_лет, более_2_лет).
        
Признаки в датасете `'ferma_main.csv'` можно разделить на группы.
- Характеристики коровы: `'id'`, `'Порода'`, `'порода папы_быка'`, `'Возраст'`.
- Характеристики корма: `'ЭКЕ (Энергетическая кормовая единица)'`, `'Сырой протеин, г'`, `'СПО (Сахаро-протеиновое соотношение)'`.
- Характеристика пастбища: `'Тип пастбища'`.
- Характеристики молока: `'Удой, кг'`, `'Жирность,%'`, `'Белок,%'`, `'Вкус молока'`.

Такое разделение помогает понять, как взаимосвязаны признаки в датасете.

Описание данных:

- `'Текущий_уровень_белок,%'` — содержание белков в молоке (в процентах).
- `'Возраст'` — возраст коровы, бинарный признак (менее_2_лет, более_2_лет).
        
Данные в файлах `'ferma_main.csv'` и `'cow_buy.csv'` похожи, но у датасета `'cow_buy.csv'` несколько особенностей:
- Содержание белков и жиров в молоке указано на момент продажи — когда «ЭкоФерма» давала коровам свой корм.
- Параметры корма `'ЭКЕ (Энергетическая кормовая единица)'`, `'Сырой протеин, г'` и `'СПО (Сахаро-протеиновое соотношение)'` отсутствуют. Технологи заказчика пересмотрели подход к кормлению: для новых коров планируется увеличить значения каждого из этих параметров на 5%.

```python
plt.figure(figsize=(16,10))

sns.heatmap(ferma_main.drop('id', axis=1).corr(), annot=True, fmt='.2f'); # матрица корреляции

plt.title('Матрица корреляции ferma_main', fontsize=20)

plt.show()
```

Годовой удой имеет высокую тесноту связи с сахаро-протеиновым соотношением (0.81, это наиболее тесная из связей для целевого признака) и энергетической кормовой единицей (0.77), заметную связь с жирностью молока (0.6; вероятно, из-за наличия зависимостей от одних и тех же параметров), умеренную - с сырым протеином (0.45). Связь с белком молока отсутствует.

Белок в молоке выявил лишь слабую линейную связь (0.23) с содержанием сырого протеина в корме коровы (что вполне логично, ведь белок из корма вполне может переходить в готовый продукт, но не до конца). Связь с СПО, ЭКЕ, жирностью молока и годовым удоем (то есть остальными числовыми параметрами) отсутствует. Возможно, она существует, но не является линейной.

Жирность молока выявила заметную линейную зависимость от энергетической кормовой единицы (0.64), сахаро-протеинового соотношения (0.63), и ранее указанную связь с годовым удоем (0.6) ввиду общих параметров, от которых они (жирность и удой) зависят. Также видна умеренная связь с содержанием сырого протеина в корме (0.36). Связь с белком молока отсутствует.

СПО показало высокую тесноту связи с ЭКЕ (0.75) и ранее уупомянутую с годовым удоем (0.81), также ранее упомянутую заметную с жирностью молока (0.63), и умеренную - с сырым протеином в корме (0.45). Связь с белком молока отсутствует.

Сырой протеин показывает умеренную тесноту связи с СПО (0.45), годовым удоем (также 0.45), ЭКЕ (0.39) и жирностью молока (0.36), слабую - с белком молока (0.23).

ЭКЕ показывает высокую тесноту связи с годовым удоем (0.77) и СПО (0.75), заметную - с жирностью молока (0.64), умеренную - с сырым протеином (0.39). Связь с белком молока отсутствует.

Однако, корреляция Пирсона хорошо подходит для тех признаков, которые распределены нормально. Но наши гистограммы показали, что такое распределение характерно не для всех. К тому же, у нас остаются "неучтенными" категорийные признаки.

Библиотека phik отсутствует в нашем модуле, поэтому применить коэфициент корреляции Фи как наиболее оптимальный для нашей задачи не выйдет.

Имеется еще один вариант: ранговая корреляция Спирмена, подходящая для порядковых или непрерывных данных, которые нарушают предположения о корреляции Пирсона.

Поэтому попробуем теперь применить корреляцию Спирмена, чтобы измерить связь между признаками более корректно.

```python
plt.figure(figsize=(16,10))

sns.heatmap(ferma_main.drop('id', axis=1).corr(method='spearman'), annot=True, fmt='.2f'); # матрица корреляции

plt.title('Матрица корреляции ferma_main', fontsize=20)

plt.show()
```

Наиболее высокую связь с годовым удоем показали СПО (коэфициент корреляции Спирмена составил 0.79) и ЭКЕ (0.77), среднюю - жирность молока (0.69) и сырой протеин (0.44), белок молока корреляцию не обнаружил.

Белок в молоке проявил слабую связь с сырым протеином (0.23) и очень слабую - с СПО (0.11) (похоже, его уровень больше зависит не от корма, а от иных характеристик коровы, либо корма - но неизвестных нам параметров).

Жирность молока проявила сильную связь с СПО (0.70), годовым удоем и ЭКЕ (0.69), среднюю - с сырым протеином (0.39).

Сахаро-протеиновое соотношение показало сильную связь с годовым удоем (0.79), ЭКЕ (0.74) и жирностью молока (0.70), среднюю - с сырым протеином (0.49, что выглядит немного странным, ввиду схожести названий данных признаков) и очень слабую (0.11) с белком в молоке.

Сырой протеин проявил средней силы связи с СПО (0.49), годовым удоем (0.44), жирностью молока (0.39) и ЭКЕ (0.38), слабую - с белком в молоке (0.23).

Энергетическая кормовая единица проявила сильную связь с годовым удоем (0.77), СПС (0.74), жирностью молока (0.69), среднюю - с сырым протеином (0.38).

В целом, в нашем случае признаки на матрице корреляции Спирмена ведут себя похожим образом с матрицей корреляции Пирсона, но определяющие зависимость коэфициенты рознятся.

Хорошая новость - для модели годового удоя можно использовать линейную регрессию. Плохая - из-за связей параметров друг с другом может потребоваться часть из них убрать. С другой стороны, ни один из коэфициентов не попадает в диапазон от 0.9 до 0.95, а значит говорить о мультимодальности нельзя.

***Выводы***

Для изучния степени зависимости признаков друг с другом были построены матрица корреляции по Пирсону (как болнее подходящая для линейной зависимости) и по Спирмену (как более корректный вариант из доступных для видимых ранее распределений признаков, который мы и приняли за основной).

Один из целевых признаков (годовой удой) выявил различной тесноты связи с остальными параметрами (за исключением процента белка в молоке): высокую с СПО (коэффициет корреляции по Спирмену составляет 0.79) и ЭКЕ (0.77), заметную с жирностью молока (0.69), умеренную - с сырым протеином (0.44).

Также матрица корреляции показала множество разной тесноты <s>линейных</s> связей между различными параметрами (за исключением белка, который продемонстрировал лишь слабую связь (0.23) с содержанием сырого протеина в корме коровы и очень слабую - с СПО (0.11)).

В нашем случае признаки на матрице корреляции Спирмена вели себя похожим образом с матрицей корреляции Пирсона, но определяющие зависимость коэфициенты рознятся.

Для модели годового удоя можно использовать линейную регрессию, однако из-за связей параметров друг с другом потенциально придется часть из них убрать.

```python
# разделяем целевой и входящие
def target_sep(df, target):
    try:
        X = df.drop(target,axis=1)
        y = df[target]
        return X, y
    except:
        pass
```

Разделение признаков через функцию прошло успешно.

Обозначим наши категориальные и числовые признаки.

```python
# функция для разделения колонок по типу содержимого
def num_cat_sort(df):

    cat_col_names = [] # список имен категориальных
    num_col_names = [] # список имен числовых столбцов

    col_list = df.columns # список имен всех колонок

    try:
        for col in col_list:
            if df[col].dtypes == 'int' or df[col].dtypes == 'float':
                num_col_names.append(col)

            elif df[col].dtypes == 'object' or df[col].dtypes == 'category':
                cat_col_names.append(col)

            else:
                pass
    except:
        pass

#    return  col_list, num_col_names, cat_col_names
    return num_col_names, cat_col_names
```

```python
num_col_names, cat_col_names = num_cat_sort(X) # получаем списки имен столбцов
```

Теперь приступим к подготовки данных к обучению: разделим данные на тренировочные и тестовые, масштабируем количественные и закодируем категориальные.

Для удобства и возможности использования данного кода в построении и других моделей "упакуем" его в функцию.

```python
# подготовка данных для модели
def spl_trans_learn(X, y):

    #Разделим на тестовую и тренировочную выборку
    X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    random_state = RANDOM_STATE
    )

    # Масштабируем количественные признаки.
    scaler = StandardScaler() # определяем метод масштабирования
    # обучаем и трансформируем количественных признаки из тренировочной выборки в одну команду
    X_train_scaled = scaler.fit_transform(X_train[num_col_names])
    # преобразовываем количественные признаки в тестовой выборке
    X_test_scaled = scaler.transform(X_test[num_col_names])

    # Кодируем категориальные переменные.
    encoder = OneHotEncoder(drop='first', sparse=False) #  выбор метода кодирования, его инициализация
    # обучение и преобразование категориальных признаков из тренировочной выборки в одну команду
    X_train_ohe = encoder.fit_transform(X_train[cat_col_names])
    # получим от кодировщика и сохраним в переменную названия новых столбцов
    encoder_col_names = encoder.get_feature_names()
    # преобразовываем категориальные признаки в тестовой выборке
    X_test_ohe = encoder.transform(X_test[cat_col_names])

    # Создадаем тренировочный и тестовый датафрейм из закодированных и отмасштабированных данных.
    X_train_ohe = pd.DataFrame(X_train_ohe, columns=encoder_col_names) # тренировочные ohe
    X_test_ohe =  pd.DataFrame(X_test_ohe, columns=encoder_col_names) # тестовые ohe
    X_train_scaled = pd.DataFrame(X_train_scaled, columns=num_col_names) # тренировочные scaled
    X_test_scaled = pd.DataFrame(X_test_scaled, columns=num_col_names) # тестовые scaled

    # Объединяем все преобразованные данные в тренировочный и тестовый датасет
    # и сбрасываем индексы в датафрейме с числовыми индексами, чтобы избежать ошибок
    X_train = pd.concat([X_train_ohe, X_train_scaled], axis = 1)
    X_test = pd.concat([X_test_ohe, X_test_scaled], axis = 1)

    return X_train, X_test, y_train, y_test, encoder, scaler
```

```python
X_train_lrf, X_test_lrf, y_train_lrf, y_test_lrf, encoder_lrf, scaler_lrf = spl_trans_learn(X, y)
```

Теперь можно обучить модель. Также создадим функцию для возможности многоразового ее использования в будующем.

```python
# модель линейной регрессии
def lin_reg(X_train, y_train):
    model_lr = LinearRegression() # инициализируем модель линейной регрессии
    model_lr.fit(X_train, y_train) # обучаем модель на тренировочных данных
    return model_lr
```

```python
# рассчет коэфициента детеминации
r2_lrf = r2_score(y_test_lrf, predictions_lrf)
r2_lrf
```

- MAE(средняя абсолютная ошибка).

```python
# подготовка данных для модели
def trans_pres(X, encoder, scaler):

    # преобразовываем количественные признаки в выборке для предсказания
    X_scaled = scaler.transform(X[num_col_names])

    # получим от кодировщика и сохраним в переменную названия новых столбцов
    encoder_col_names = encoder.get_feature_names()
    # преобразовываем категориальные признаки в выборке для предсказания
    X_ohe = encoder.transform(X[cat_col_names])

    # Создадаем датафреймы из закодированных и отмасштабированных данных.
    X_ohe = pd.DataFrame(X_ohe, columns=encoder_col_names) # тестовые ohe
    X_scaled = pd.DataFrame(X_scaled, columns=num_col_names) # тестовые scaled

    # Объединяем все преобразованные данные в один датасет
    # и сбрасываем индексы в датафрейме с числовыми индексами, чтобы избежать ошибок
    X = pd.concat([X_ohe, X_scaled], axis = 1)

    return X
```

#### Вариант 2

Подберем порог с помощью перебора. Не самый удобный и наглядный вариант, зато надежный.

Метрика precision показывает точность, с которой модель присваивает объектам класс 1, то есть не слишком ли часто модель выставляет класс 1 объектам класса 0. Диапазон (0.5; 1] являлся бы подходящим для того, чтобы его перебирать в поисках подходящего значения.

Однако мы упростили себе задачу, сузив ранее диапазон возможных значений [0.82, 0.91).

Для линейной регрессии r^2 (коэфициент детерминации), MSE (среднеквадратичная ошибка), RMSE (корень среднеквадратичной ошибки), и MAE(средняя абсолютная ошибка), а также анализ остатков. Каждая из них по-своему хороша и показывает лучший результат в определенной ситуации (например, MAE менее чувствительна к выбросам), какие-то удобнее в интерпритации (та же MAE указывает ошибку в тех же единицах, что и сам параметр). Поэтому важно знать их все для объективной оценки модели.

Для задач логистической регрессии используются метрики accuracy (показывает долю верных ответов модели), recall (точность присваивания объектам класса 1), precision (полнота присваивания объектам класса 1), а также построение матрицы ошибок для наглядного отображения ошибок первого и второга класса. В зависимости от целей, которые ставит заказчик, белее важной может быть либо метрика precision (снизить попадание объекта класса 0 в класс 1), так и recall (снизить попадание объектов класса 1 в класс 0). в нашем случае ключевой метрикой оказалась  precision.

Мы получили модель с низким качеством из-за недостатка данных. Наверняка, факторов влияющих на вкус больше, чем на удой. Например, может иметь значение то, какую именно траву ела пасущаяся корова: согласно открытым источникам поедание коровой  пижмы или полыни придает молоку горьковатый вкус, который вряд ли можно считать вкусным. Также горечь молоку могут придать заболевания коровы, например - маститы. Возможно, появление новых отслеживаемых признаков позволит улучшить модель в будующем.
